```python
train_sse = np.sum((train_predicted - y_train)**2, axis = 0)
train_sse
```

```python
train_sse = []
test_sse = []

tr_size = 0.1

while tr_size <= 0.9:
    x_train, x_test, y_train, y_test = train_test_split(x, y, train_size = tr_size, random_state=1)
    
    mod = LinearRegression()
    mod.fit(x_train, y_train)
    
    pred_train = mod.predict(x_train)
    pred_test = mod.predict(x_test)
    
    tr_sse = np.sum((pred_train - y_train)**2, axis = 0)
    te_sse = np.sum((pred_test - y_test)**2, axis = 0)
    
    train_sse.append(tr_sse)
    test_sse.append(te_sse)
    
    tr_size += 0.1
```

```python
trainSize = []
for i in range(1, 10):
    trainSize.append(i/10)

plt.title("Effect of splitting dataset on Training and Testing Error")
plt.plot(trainSize, np.array(train_sse), c = 'r')
plt.plot(trainSize, np.array(test_sse), c = 'b')
plt.grid(True)
plt.xlabel("Train_Size")
plt.ylabel("SSE")

plt.legend(["Training Error", "Testing Error"])
```

```python
features = ['OCT-DEC', 'JAN-FEB', 'MAR-MAY']
X = v[features]
y = v['ANNUAL']
```

```python
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)
```

```python
train_mse = mean_squared_error(y_train, y_train_pred)
test_mse = mean_squared_error(y_test, y_test_pred)
```

```python
fig = plt.figure(figsize=(12, 6))

# Training data plot
ax_train = fig.add_subplot(121, projection='3d')
ax_train.scatter(X_train['OCT-DEC'], X_train['JAN-FEB'], y_train, color='blue', label='Actual')
ax_train.scatter(X_train['OCT-DEC'], X_train['JAN-FEB'], y_train_pred, color='red', label='Predicted', alpha=0.5)
ax_train.set_xlabel('OCT-DEC')
ax_train.set_ylabel('JAN-FEB')
ax_train.set_zlabel('ANNUAL')
ax_train.set_title(f'Training Data\nMSE: {train_mse:.2f}')
ax_train.legend()

# Testing data plot
ax_test = fig.add_subplot(122, projection='3d')
ax_test.scatter(X_test['OCT-DEC'], X_test['JAN-FEB'], y_test, color='blue', label='Actual')
ax_test.scatter(X_test['OCT-DEC'], X_test['JAN-FEB'], y_test_pred, color='red', label='Predicted', alpha=0.5)
ax_test.set_xlabel('OCT-DEC')
ax_test.set_ylabel('JAN-FEB')
ax_test.set_zlabel('ANNUAL')
ax_test.set_title(f'Testing Data\nMSE: {test_mse:.2f}')
ax_test.legend()

plt.show()
```

```python
plt.figure(figsize=(8, 6))
errors = {'Training Error': train_mse, 'Testing Error': test_mse}
plt.bar(errors.keys(), errors.values(), color=['blue', 'orange'])
plt.ylabel('Mean Squared Error')
plt.title('Training and Testing Errors')
plt.show()
```

```python
features = ['OCT-DEC', 'JAN-FEB', 'MAR-MAY']
X = v[features]
y = v['ANNUAL']

train_sse = []
test_sse = []
```

```python
train_size = 0.1
```

```python
while train_size <= 0.9:
    # Splitting the data
    X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=train_size, random_state=1)
    
    # Training the model
    model = LinearRegression()
    model.fit(X_train, y_train)
    
    # Predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)
    
    # Calculating SSE
    train_sse.append(np.sum((y_train_pred - y_train) ** 2))
    test_sse.append(np.sum((y_test_pred - y_test) ** 2))
    
    # Increment the train size
    train_size += 0.1
```

```python
train_sizes = [i / 10 for i in range(1, 10)]

# Plotting the SSE for training and testing
plt.figure(figsize=(10, 6))
plt.title("Effect of Splitting Dataset on Training and Testing Error")
plt.plot(train_sizes, train_sse, color='red', marker='o')
plt.plot(train_sizes, test_sse, color='blue', marker='o')
plt.grid(True)
plt.xlabel("Train Size")
plt.ylabel("Sum of Squared Errors (SSE)")
plt.legend(["Training Error", "Testing Error"])
plt.show()
```
