```python
subset = df.loc[ : , ['Page total likes','Type','Post Month', 'like','share','comment']]
print("\n Subset values :\n",subset.head())
```

```python
merged_df = pd.DataFrame({'Type' : ['Photo','Status'],'Category' :['Visual','Textual']})
merge = pd.merge(subset,merged_df,on='Type',how = 'left')
print("\n merged values :\n",merge.head())
```

```python
sorted_df = merge.sort_values(by = 'like',ascending = False)
print("\n Sorted values :\n",sorted_df.head())
```

```python
print("\n Subset values :\n",sorted_df.head(6).T)
```

```python
melted = pd.melt(subset.head(),id_vars = ['Type'])
print("\n Melted values :\n",melted.head())
```

```python
## ii. Pivot (Casting): Long → Wide
casted = melted.pivot_table(index='Type', columns='variable', values='value', aggfunc='first')
print(casted)
```
