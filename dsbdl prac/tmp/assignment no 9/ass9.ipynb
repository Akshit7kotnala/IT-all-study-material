```python
import requests
from bs4 import BeautifulSoup as bs
import pandas as pd


# Soup is a python library used for scrapping websites 
```

```python
link = "https://www.flipkart.com/apple-iphone-16-pro-max-black-titanium-256-gb/p/itm7e75db4f27bd5?pid=MOBH4DQFNXH8SZ9D&lid=LSTMOBH4DQFNXH8SZ9DUXE3QP&marketplace=FLIPKART&q=iphone+16+pro+max&store=tyy%2F4io&srno=s_1_1&otracker=AS_QueryStore_OrganicAutoSuggest_1_7_na_na_na&otracker1=AS_QueryStore_OrganicAutoSuggest_1_7_na_na_na&fm=organic&iid=582126a7-83f5-48ad-8c9a-66a8bc3008a0.MOBH4DQFNXH8SZ9D.SEARCH&ppt=pp&ppn=pp&ssid=vnp5o80o7k0000001745389568970&qH=7af9e8ffea95ab61"
```

```python
page = requests.get(link)

page
```

```python
soup = bs(page.content,'html.parser')
# parses HTML of the page

print(soup.prettify())
```

```python
names = soup.find_all('p',class_='_2NsDsF AwS1CA')
# ( html tag , class)

names
```

```python
cust_name = []

for i in range(0,len(names)):

    cust_name.append(names[i].get_text())

cust_name
```

```python
title = soup.find_all('p',class_='z9E0IG')

title
```

```python
buyer_type = soup.find_all('p',class_='MztJPv')

buyer_type
```

```python
rating_divs = soup.find_all('div', class_='XQDdHH Ga3i8K')

ratings = []
for div in rating_divs:
    rating_text = div.get_text(strip=True)
    ratings.append(int(rating_text[0]))  # Extract only the digit

ratings
```

```python
review = soup.find_all('div',class_='ZmyHeo')

review
```

```python
cust_add = []

for i in range(0,len(buyer_type)):

    cust_add.append(buyer_type[i].get_text())

cust_add
```

```python
cust_rev = []

for div in review:
    comment_div = div.find('div').find('div')  # Access nested div with comment ( theres two nested divs)
    if comment_div:
        cust_rev.append(comment_div.get_text(separator=" ", strip=True)) # strip white-spaces

cust_rev
```

```python
df = pd.DataFrame()
```

```python
df['Customer Name']=cust_name
```

```python
df['Review title']=title
df['Customer Address']=cust_add
df['cutometer Review']=cust_rev
df['rating']=ratings
```
